```python
import matplotlib.pyplot as plt
import pandas as pd
course_enroll = {
    "courses": ["java", "python", "c++", "javascript"],
    "enrolled": [70, 80, 90, 60],
}

course_df =pd.DataFrame(course_enroll)

plt.bar(course_df["courses"], course_df["enrolled"], color="blue")
plt.xlabel("Course")
plt.ylabel("Enrollment")
plt.title("Course Enrollment")
plt.bar_label("Courses enrolled student",)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

weekly_sales = {
    "days": ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"],
    "first_week": [100, 150, 200, 180, 250],
    "second_week": [120, 170, 220, 200, 270],
}

sales_df = pd.DataFrame(weekly_sales)

plt.plot(sales_df["days"], sales_df["first_week"], label="First Week", marker="o")
plt.plot(sales_df["days"], sales_df["second_week"], label="Second Week", marker="o")
plt.title
plt.title("shop")
plt.title
plt.xlabel("Day")
plt.ylabel("Sales")
plt.legend()
plt.show()

```
