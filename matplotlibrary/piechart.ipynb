```python
import matplotlib.pyplot as plt

phone_names = ["iPhone 16", "Samsung Galaxy S25", "Google Pixel 9", "OnePlus 13"]
sales = [35, 30, 20, 15]  # phones sold

plt.pie(sales, labels=phone_names, autopct="%1.1f%%", startangle=90)
plt.title("Mobile Phone Sales")
plt.axis("equal")
plt.title
plt.show()
```

```python
### Histogram
import matplotlib.pyplot as plt


temperatures = [24, 24, 26, 26, 26, 26, 26, 26, 24, 23, 22, 22, 22, 21, 21]


plt.figure(figsize=(8, 5))
plt.hist(temperatures, bins=[20.5, 21.5, 22.5, 23.5, 24.5, 25.5, 26.5], 
         edgecolor='black', color='skyblue', rwidth=0.85)

plt.title("Distribution of Hourly Temperatures in Bengaluru (Oct 23, 2026)", fontsize=14)
plt.xlabel("Temperature (°C)", fontsize=12)
plt.ylabel("Frequency (Number of Hours)", fontsize=12)
plt.grid(axis='y', alpha=0.75, linestyle='--')
plt.xticks([21, 22, 23, 24, 25, 26])
plt.show()
```
